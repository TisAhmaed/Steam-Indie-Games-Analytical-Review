### Import the data from csv files



```python
import pandas as pd
import numpy as np

#Import titanic dataset
df=pd.read_csv("games_raw.csv")
df.head()
```

## Get Info

# Check For Missing Values

```python
df.isnull().sum()
```

```python
# drop games whose review totals didn't come back
df = df.dropna(subset=["total_positive", "total_negative"])
```

## Remove null Values From Data

```python
df['genres'] = df['genres'].fillna("RPG")
```

```python
df[df['genres'].isna()]
```

## Check For Duplicate Rows

```python

df.duplicated().sum()
```

## Normalizing Data

Datetime Normalization

```python
df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce")
```

## Boolean Normalization

```python
df["is_free"] = df["is_free"].astype(bool)
```

```python
df
```

## Normalization of Owners

```python
def average_owners(owner_range):
    # SteamSpy sometimes returns nothing -> leave it missing
    if pd.isna(owner_range):
        return np.nan
    low, high = owner_range.split(" .. ")
    low = int(low.replace(",", ""))
    high = int(high.replace(",", ""))
    return (low + high) / 2

df["owners_avg"] = df["owners"].apply(average_owners)
```

```python
df
```
